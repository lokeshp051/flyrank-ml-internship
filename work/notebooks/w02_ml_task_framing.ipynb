## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This is primarily a ranking/scoring task rather than a pure classification problem.
Instead of predicting a simple yes/no outcome for each page, the goal is to generate an ordered priority list based on multiple signals such as impressions, freshness, CTR, and trend direction. Reviewers then use this ranking to decide which pages to review first, making the output a decision-support tool rather than a final decision.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target/proxy: is_declining_label = (trend_direction == "down") — a proxy, not a perfect target, since it's calculated from the current window rather than a true future outcome.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Success metric: Precision@50 — the percentage of the top 50 prioritized pages that are actually labeled as declining.
This metric reflects the real workflow, where reviewers have limited time and can only inspect a small number of pages.
A higher Precision@50 means more reviewer effort is spent on genuinely valuable refresh opportunities.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

My hand rule hit Precision@20 = 0.900 but only Precision@50 = 0.680. A depth-3 tree matched it at the top but reached Precision@50 = 0.720 — a fixed rule runs out of nuance past the obvious cases; a model weighs more signals together, which matters deeper in the list.
